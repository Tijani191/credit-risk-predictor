```python
# Separate features and target
X = df.drop('Class', axis=1)
y = df['Class']
```

```python
print(f"Features shape: {X.shape}")
print(f"Target shape: {y.shape}")
print(f"Target class balance: {y.value_counts()}")
```

```python
#Train-Test Split (Before Balancing!)
```

```python
from sklearn.model_selection import train_test_split
# 70-30 split
X_train, X_test, y_train, y_test = train_test_split(
 X, y, test_size=0.30, random_state=42, stratify=y
)
```

```python
print(f"Test set: {X_test.shape[0]} samples")
```

```python
print(f"Test fraud %: {y_test.mean()*100:.2f}%")
```

```python
#Handle Class Imbalance with SMOTE 
```

```python
#Apply SMOTE to Training Set Only
from imblearn.over_sampling import SMOTE, ADASYN
```

```python
# SMOTE: oversample minority (fraud) class
smote = SMOTE(random_state=42, k_neighbors=5)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)
```

```python
print(f"Original train set class distribution:")
print(y_train.value_counts())
```

```python
print(f"\nAfter SMOTE:")
print(pd.Series(y_train_balanced).value_counts())
```

```python
# Now both classes are roughly equal
```

```python
# Alternative: ADASYN (adaptive)
adasyn = ADASYN(random_state=42)
X_train_balanced_ada, y_train_balanced_ada = adasyn.fit_resample(X_train, y_train)
# ADASYN weights synthetic samples by difficulty — slightly better but slower
```

```python
#Visualise the Effect
```

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
import joblib
import pandas as pd

# 1. Initialize the API
app = FastAPI(
    title="Credit Card Fraud Detection API", 
    description="API for predicting fraudulent transactions using XGBoost and SMOTE."
)

# 2. Load the trained XGBoost model
try:
    model = joblib.load('xgboost_fraud_model.joblib')
except Exception as e:
    print(f"Error loading model: {e}")

# 3. Define the strict input data schema
class TransactionData(BaseModel):
    Time: float
    V1: float
    V2: float
    V3: float
    V4: float
    V5: float
    V6: float
    V7: float
    V8: float
    V9: float
    V10: float
    V11: float
    V12: float
    V13: float
    V14: float
    V15: float
    V16: float
    V17: float
    V18: float
    V19: float
    V20: float
    V21: float
    V22: float
    V23: float
    V24: float
    V25: float
    V26: float
    V27: float
    V28: float
    Amount: float

# 4. Create the prediction endpoint
@app.post("/predict")
def predict_fraud(transaction: TransactionData):
    try:
        # Convert the incoming JSON payload into a pandas DataFrame
        # The column names must match the names used during XGBoost training
        data_dict = transaction.dict()
        features = pd.DataFrame([data_dict])
        
        # Make a prediction (0 = Legitimate, 1 = Fraud)
        prediction = model.predict(features)
        
        # Extract the probability score of the transaction being fraudulent
        probability = model.predict_proba(features)[0][1]
        
        return {
            "prediction": int(prediction[0]),
            "fraud_probability": float(probability),
            "status": "Fraudulent" if prediction[0] == 1 else "Legitimate"
        }
        
    except Exception as e:
        raise HTTPException(status_code=400, detail=f"Prediction error: {str(e)}")
```
